## 1. Conexión y carga digital
*Crea una conexión persistente a una base de datos DuckDB llamada viajes.duckdb y carga los ficheros Parquet creando las tablas correspondientes (paises, lugares_interes, clientes y reservas) si no existen previamente.*

```python
import duckdb
import pandas as pd
from pyarrow import parquet
import pyarrow

# Conexión persistente
bd = duckdb.connect("viajes.duckdb")

t_clientes = parquet.read_table("viajes/clientes.parquet")
t_paises = parquet.read_table("viajes/paises.parquet")
t_lugares_interes = parquet.read_table("viajes/lugares_interes.parquet")
t_reservas = parquet.read_table("viajes/reservas.parquet")

# Carga de ficheros parquet
bd.sql("CREATE TABLE IF NOT EXISTS clientes as SELECT * FROM t_clientes")
bd.sql("CREATE TABLE IF NOT EXISTS paises as SELECT * FROM t_paises")
bd.sql("CREATE TABLE IF NOT EXISTS lugares_interes as SELECT * FROM t_lugares_interes")
bd.sql("CREATE TABLE IF NOT EXISTS reservas as SELECT * FROM t_reservas")

bd.sql("SHOW TABLES")
```
